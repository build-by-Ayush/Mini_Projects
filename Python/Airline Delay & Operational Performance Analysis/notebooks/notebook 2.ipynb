```python
import pandas as pd
df = pd.read_csv('../data/flights_sample_3m.csv')
```

```python
print(df.columns)
```

```python
time_cols = [
    'CRS_DEP_TIME',
    'DEP_TIME',
    'WHEELS_OFF',
    'WHEELS_ON',
    'CRS_ARR_TIME',
    'ARR_TIME',
]

count_2400 = (df[time_cols] == 2400).sum()
print('Rows with 2400:')
print(count_2400)
```

```python
# Check the 5 lowest and 5 highest unique values for each time column
for col in time_cols:
  print(f'--- {col} ---')
  print('Lowest values:', sorted(df[col].dropna().unique())[:5])
  print('Highest values:', sorted(df[col].dropna().unique())[-5:])
```

```python
invalid_times = {}

for col in time_cols:
  s = df[col].dropna()

  invalid_mins = (s % 100) > 59

  invalid_hrs = ((s // 100) > 23) & (s != 2400)

  negative_vals = s < 0

  total_invalid = s[invalid_mins | invalid_hrs | negative_vals]
  invalid_times[col] = total_invalid.unique()

  print(f'{col}: {len(total_invalid)} invalid entries found.')
  if len(total_invalid) > 0:
    print(f'   Unique invalid values: {total_invalid.unique()}\n')
```
