```python
import pandas as pd

# 1. Grab all 5 columns that start with 'DELAY_DUE_'
delay_cols = [col for col in df.columns if col.startswith('DELAY_DUE_')]

# 2. Create boolean masks
# Mask for rows where AT LEAST ONE delay column is not null
at_least_one_delay = df[delay_cols].notna().any(axis=1)

# Mask for rows where ALL FIVE delay columns are null
all_delays_null = df[delay_cols].isna().all(axis=1)

# 3. Print the row counts
print(f"1. Rows with at least one delay-cause value: {at_least_one_delay.sum()}")
print(f"2. Rows with all five delay-cause values null: {all_delays_null.sum()}")

# Create a subset of rows that actually have delay data
df_with_delays = df[at_least_one_delay]

# Check how many of these flights were CANCELLED or DIVERTED
print("--- Cancellation & Diversion Behavior ---")
print("Cancelled status counts:")
print(df_with_delays['CANCELLED'].value_counts(dropna=False))

print("\nDiverted status counts:")
print(df_with_delays['DIVERTED'].value_counts(dropna=False))

print("\n--- Arrival Delay (ARR_DELAY) Statistics ---")
print(df_with_delays['ARR_DELAY'].describe())
```
